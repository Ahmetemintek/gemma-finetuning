```python
import os

# Validate Google Drive mount
print("🔍 Validating Google Drive connection...\n")

# Check if drive is mounted
drive_path = '/content/drive/MyDrive'
if os.path.exists(drive_path):
    print("✅ Google Drive successfully mounted!")
    print(f"📂 Drive path: {drive_path}\n")
    
    # List top-level directories
    print("📁 Top-level directories in MyDrive:")
    dirs = [d for d in os.listdir(drive_path) if os.path.isdir(os.path.join(drive_path, d))]
    for d in dirs[:10]:  # Show first 10
        print(f"  - {d}")
    if len(dirs) > 10:
        print(f"  ... and {len(dirs) - 10} more")
    
    # Test write permissions
    print("\n🔧 Testing write permissions...")
    test_file = os.path.join(drive_path, '.colab_test.txt')
    try:
        with open(test_file, 'w') as f:
            f.write('Connection test successful')
        os.remove(test_file)
        print("✅ Write permissions confirmed!")
    except Exception as e:
        print(f"❌ Write test failed: {e}")
    
    print("\n✨ Google Drive is ready to use!")
else:
    print("❌ Drive not mounted at /content/drive/MyDrive")
    print("   Please run the mount cell above first.")
```
